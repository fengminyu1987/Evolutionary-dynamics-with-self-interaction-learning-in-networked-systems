```python
import matplotlib.pyplot as plt
import networkx as nx
def draw_graph(G, node_options, edge_options, edge_labels=None):
    pos = nx.spring_layout(G, iterations=10, seed=20000)
    # Draw the nodes with the specified options    
    nx.draw_networkx_nodes(G, pos, **node_options)
    # Draw the node labels with specified font properties    
    nx.draw_networkx_labels(G, pos, font_size=14, font_color=label_color, font_weight='bold')
    # Draw the edges with the specified options    
    nx.draw_networkx_edges(G, pos, **edge_options)
    # Draw edge labels if they are provided    
    if edge_labels:
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels,connectionstyle="arc3, rad=0.3",font_size=12,font_color="orange",bbox={"alpha": 1, "color": "white"})
    # Turn off the axis and ensure the graph aspect ratio is equal    
    plt.axis('off')    
    plt.gca().set_aspect('equal')
    # Display the graph    
    plt.show()
plt.figure(figsize=(8,8))
N=7
edge_options = {'edge_color': 'b','width': 2.5,'alpha': 0.7,'connectionstyle': 'arc3, rad = 0.3','arrowstyle': '-|>','arrowsize': 20}
graph_color = "#4986e8"
label_color = "#ffffff"
node_options = {'node_color': graph_color,'node_size': 1000,'edgecolors': 'black','linewidths': 2}
G=nx.star_graph(N)
draw_graph(G, node_options, edge_options)
```

```python
#Star
import numpy as np
def star(N,p, q):
    # 分子部分
    numerator = 3 * (1 + p) * (-1 + N + q) * (
        -(8/3) 
        + N**3 * (4/3 + (11 * p)/3 + p**2) 
        + (14 * q)/3 
        - (4 * q**2)/3 
        + p**2 * (-2 + (8 * q)/3) 
        + p * (-6 + 10 * q - (8 * q**2)/3) 
        + N**2 * (-(16/3) + 3 * q + p**2 * (-4 + (4 * q)/3) + p * (-(40/3) + 7 * q)) 
        + N * (20/3 + p**2 * (5 - 4 * q) - (23 * q)/3 + (4 * q**2)/3 + p * (47/3 - 17 * q + (8 * q**2)/3))
    )

    # 分母部分
    denominator = 2 * (
        q * (-N**3 + N * (-1 - q/2) + N**2 * (2 - q/2) + q) 
        + p**3 * (2 + N**4 - 5 * q + 4 * q**2 + N**3 * (-5 + (5 * q)/2) + N * (-7 + (25 * q)/2 - 6 * q**2) + N**2 * (9 - 10 * q + 2 * q**2)) 
        + p * (4 + N**4 - 11 * q + 12 * q**2 - 2 * q**3 + N**3 * (-7 + (5 * q)/2) + N**2 * (15 - 16 * q + 5 * q**2) + N * (-13 + (49 * q)/2 - 17 * q**2 + 2 * q**3)) 
        + p**2 * (8 + 4 * N**4 - 22 * q + 21 * q**2 - 4 * q**3 + N**3 * (-20 + 12 * q) + N**2 * (36 - 46 * q + (27 * q**2)/2) + N * (-28 + 56 * q - (69 * q**2)/2 + 4 * q**3))
    )

    # 返回分子/分母的值
    return numerator / denominator
```

```python
#hub-hub joined star
def star_HH(N,alpha,gamma):
# 定义分子部分
    numerator = (
        3 * (1 + alpha) * (N + gamma) * (
            N**4 * (alpha**3 + (16/3) * alpha**2 + (25/3) * alpha + (10/3)) +
            N**3 * (
                alpha**3 * ((4/3) * gamma - 2) +
                alpha**2 * (10 * gamma - (25/3)) +
                alpha * (20 * gamma - (34/3)) +
                (29/3) * gamma - (13/3)
            ) +
            N**2 * (
                alpha**3 * (-(7/3) * gamma + (1/3)) +
                alpha**2 * ((11/3) * gamma**2 - (38/3) * gamma + 1) +
                alpha * ((40/3) * gamma**2 - (64/3) * gamma + 1) +
                (26/3) * gamma**2 - 9 * gamma + (1/3)
            ) +
            N * (
                alpha**3 * (-(1/3) * gamma + (2/3)) +
                alpha**2 * (-(10/3) * gamma**2 - gamma + 2) +
                alpha * ((8/3) * gamma**3 - (32/3) * gamma**2 - gamma + 2) +
                3 * gamma**3 - (16/3) * gamma**2 - (1/3) * gamma + (2/3)
            ) +
            gamma * (
                (4/3) * alpha**3 + 4 * alpha**2 +
                alpha * (-(5/3) * gamma**2 + 4) +
                (1/3) * gamma**3 - gamma**2 + (4/3)
            )
        )
    )

    # 定义分母部分
    denominator = (
        2 * (
            N**5 * (alpha**4 + (11/2) * alpha**3 + 9 * alpha**2 + 5 * alpha + 2) +
            N**4 * (
                alpha**4 * ((5/2) * gamma - (5/2)) +
                alpha**3 * ((33/2) * gamma - (23/2)) +
                alpha**2 * ((61/2) * gamma - 19) +
                alpha * ((33/2) * gamma - (25/2)) +
                6 * gamma - 4
            ) +
            N**3 * (
                alpha**4 * (2 * gamma**2 - (11/2) * gamma + 2) +
                alpha**3 * ((37/2) * gamma**2 - 29 * gamma + (15/2)) +
                alpha**2 * (44 * gamma**2 - (107/2) * gamma + 11) +
                alpha * (29 * gamma**2 - 36 * gamma + 7) +
                (21/2) * gamma**2 - 12 * gamma + (3/2)
            ) +
            N**2 * (
                alpha**4 * (-(7/2) * gamma**2 + 2 * gamma - (1/2)) +
                alpha**3 * gamma * ((11/2) * gamma**2 - (49/2) * gamma + (15/2)) +
                alpha**2 * ((47/2) * gamma**3 - (111/2) * gamma**2 + (23/2) * gamma + (7/2)) +
                alpha * (21 * gamma**3 - (79/2) * gamma**2 + 7 * gamma + 5) +
                9 * gamma**3 - 14 * gamma**2 + gamma + 2
            ) +
            N * (
                alpha**4 * (-(1/2) * gamma**2 + gamma - 1) +
                alpha**3 * (-5 * gamma**3 - 2 * gamma**2 + 6 * gamma - (11/2)) +
                alpha**2 * (4 * gamma**4 - (41/2) * gamma**3 - (5/2) * gamma**2 + 13 * gamma - (21/2)) +
                alpha * (6 * gamma**4 - 16 * gamma**3 - (5/2) * gamma**2 + 12 * gamma - (17/2)) +
                (7/2) * gamma**4 - (13/2) * gamma**3 - (3/2) * gamma**2 + 4 * gamma - (5/2)
            ) +
            alpha**4 * (2 * gamma**2 + 1) +
            alpha**3 * (8 * gamma**2 - (3/2) * gamma + 4) +
            alpha**2 * (-(5/2) * gamma**4 + (25/2) * gamma**2 - (9/2) * gamma + 6) +
            alpha * ((1/2) * gamma**5 - 2 * gamma**4 - (1/2) * gamma**3 + 9 * gamma**2 - (9/2) * gamma + 4) +
            (1/2) * gamma**5 - gamma**4 - (1/2) * gamma**3 + (5/2) * gamma**2 - (3/2) * gamma + 1
        )
    )

    # 最终的公式表示
    return numerator / denominator
```
